# Tarefa 1

### 1. Idade media das mães e dos pais por município (coluna munResNome)


```python
# Calcula a idade média das mães por município
media_idade_maes = sinasc_raw.groupby('munResNome')['IDADEMAE'].mean().reset_index()
media_idade_maes.columns = ['Município', 'Idade Média Mães']

# Calcula a idade média dos pais por município
media_idade_pais = sinasc_raw.groupby('munResNome')['IDADEPAI'].mean().reset_index()
media_idade_pais.columns = ['Município', 'Idade Média Pais']

# Junta as duas tabelas
media_idades = pd.merge(media_idade_maes, media_idade_pais, on='Município')

print(media_idades)
```

### 2. Peso médio dos bebes por sexo que nasceram no dia do seu aniversário por faixas de escolaridade mae
Ex: Você, aluna(o), nasceu no dia 10/01, então você precisa filtrar o conjunto de dados nessa data e calcular o peso médio dos bebês de cada sexo por faixa de escolaridade da mãe.

```python
# Converte a coluna DTNASC para o formato de data
sinasc_raw['DTNASC'] = pd.to_datetime(sinasc_raw['DTNASC'], format='%Y-%m-%d')

# Filtra os dados para o dia 15/09
data_filtrada = sinasc_raw[sinasc_raw['DTNASC'].dt.strftime('%d-%m') == '15-09']

# Calcula o peso médio dos bebês por sexo e faixa de escolaridade da mãe
peso_medio = data_filtrada.groupby(['SEXO', 'ESCMAE'])['PESO'].mean().reset_index()

print(peso_medio)
```

### 3. Qual o municipio que nasceu menos bebe em 2019?
    - qual a idade media, maxima, minima das maes nesse municipio?
    - qual a idade media, maxima, minima dos pais nesse municipio?

```python
# Encontra o município com o menor número de nascimentos
municipio_menos_nascimentos = sinasc_raw['munResNome'].value_counts().idxmin()

# Filtra os dados para o município com menos nascimentos
dados_municipio = sinasc_raw[sinasc_raw['munResNome'] == municipio_menos_nascimentos]

# Calcula as estatísticas das idades das mães
idade_maes_media = dados_municipio['IDADEMAE'].mean()
idade_maes_maxima = dados_municipio['IDADEMAE'].max()
idade_maes_minima = dados_municipio['IDADEMAE'].min()

# Calcula as estatísticas das idades dos pais
idade_pais_media = dados_municipio['IDADEPAI'].mean()
idade_pais_maxima = dados_municipio['IDADEPAI'].max()
idade_pais_minima = dados_municipio['IDADEPAI'].min()

print(f"Município com menos nascimentos: {municipio_menos_nascimentos}")
print(f"Idade média das mães: {idade_maes_media:.2f}")
print(f"Idade máxima das mães: {idade_maes_maxima}")
print(f"Idade mínima das mães: {idade_maes_minima}")
print(f"Idade média dos pais: {idade_pais_media:.2f}")
print(f"Idade máxima dos pais: {idade_pais_maxima}")
print(f"Idade mínima dos pais: {idade_pais_minima}")
```

### 4. Qual o municipio que nasceu mais bebe no mês de março?
    - qual a quantidade de filhos vivos media, maxima, minima nesse municipio?
    - qual a idade media, maxima, minima dos pais nesse municipio?



```python
# Converte a coluna DTNASC para o formato de data
sinasc_raw['DTNASC'] = pd.to_datetime(sinasc_raw['DTNASC'], format='%Y-%m-%d')

# Filtra os dados para o mês de março
dados_marco = sinasc_raw[sinasc_raw['DTNASC'].dt.month == 3]

# Encontra o município com o maior número de nascimentos em março
municipio_mais_nascimentos = dados_marco['munResNome'].value_counts().idxmax()

# Filtra os dados para o município com mais nascimentos em março
dados_municipio_marco = dados_marco[dados_marco['munResNome'] == municipio_mais_nascimentos]

# Calcula a quantidade de filhos vivos (média, máxima e mínima) nesse município
filhos_vivos_media = dados_municipio_marco['QTDFILVIVO'].mean()
filhos_vivos_maxima = dados_municipio_marco['QTDFILVIVO'].max()
filhos_vivos_minima = dados_municipio_marco['QTDFILVIVO'].min()

# Calcula a idade dos pais (média, máxima e mínima) nesse município
idade_pais_media = dados_municipio_marco['IDADEPAI'].mean()
idade_pais_maxima = dados_municipio_marco['IDADEPAI'].max()
idade_pais_minima = dados_municipio_marco['IDADEPAI'].min()

print(f"Município com mais nascimentos em março: {municipio_mais_nascimentos}")
print(f"Quantidade de filhos vivos média: {filhos_vivos_media:.2f}")
print(f"Quantidade de filhos vivos máxima: {filhos_vivos_maxima}")
print(f"Quantidade de filhos vivos mínima: {filhos_vivos_minima}")
print(f"Idade média dos pais: {idade_pais_media:.2f}")
print(f"Idade máxima dos pais: {idade_pais_maxima}")
print(f"Idade mínima dos pais: {idade_pais_minima}")
```
