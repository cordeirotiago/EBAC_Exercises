Considere a mesma base de dados utilizada na tarefa 1 ```previsao_de_renda.csv```.

1) Com o que você viu em aula, faça uma análise das variáveis qualitativas (posse de imóvel e posse de veículo) para verificar se são boas preditoras de *renda*. Alguma delas parece discriminar *renda* de forma estatisticamente significante?

````python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import ttest_ind

# Carregar a base de dados
df = pd.read_csv('previsao_de_renda.csv')

# Visualizar resumo estatístico
print(df.describe())
````

````python
# Analisar a distribuição da renda para posse de imóvel
sns.boxplot(x='posse_de_imovel', y='renda', data=df)
plt.title('Distribuição da Renda por Posse de Imóvel')
plt.show()

# Analisar a distribuição da renda para posse de veículo
sns.boxplot(x='posse_de_veiculo', y='renda', data=df)
plt.title('Distribuição da Renda por Posse de Veículo')
plt.show()

# Realizar testes estatísticos
imovel_sim = df[df['posse_de_imovel'] == 'sim']['renda']
imovel_nao = df[df['posse_de_imovel'] == 'não']['renda']
veiculo_sim = df[df['posse_de_veiculo'] == 'sim']['renda']
veiculo_nao = df[df['posse_de_veiculo'] == 'não']['renda']

# Teste t de Student para posse de imóvel
t_stat_imovel, p_value_imovel = ttest_ind(imovel_sim, imovel_nao)
print(f'Teste t para posse de imóvel: t_stat={t_stat_imovel}, p_value={p_value_imovel}')

# Teste t de Student para posse de veículo
t_stat_veiculo, p_value_veiculo = ttest_ind(veiculo_sim, veiculo_nao)
print(f'Teste t para posse de veículo: t_stat={t_stat_veiculo}, p_value={p_value_veiculo}')
````

Não. Ter veículo ou imóvel não demonstra diferença significativa, quando comparado por boxplot.

````python
df_clean = df.dropna(subset=['renda', 'posse_de_imovel', 'posse_de_veiculo'])

# Analisar a distribuição da renda para posse de imóvel com dados limpos
sns.boxplot(x='posse_de_imovel', y='renda', data=df_clean)
plt.title('Distribuição da Renda por Posse de Imóvel')
plt.show()

# Analisar a distribuição da renda para posse de veículo com dados limpos
sns.boxplot(x='posse_de_veiculo', y='renda', data=df_clean)
plt.title('Distribuição da Renda por Posse de Veículo')
plt.show()

# Realizar testes estatísticos com dados limpos
imovel_sim_clean = df_clean[df_clean['posse_de_imovel'] == 'sim']['renda']
imovel_nao_clean = df_clean[df_clean['posse_de_imovel'] == 'não']['renda']
veiculo_sim_clean = df_clean[df_clean['posse_de_veiculo'] == 'sim']['renda']
veiculo_nao_clean = df_clean[df_clean['posse_de_veiculo'] == 'não']['renda']

# Teste t de Student para posse de imóvel com dados limpos
t_stat_imovel_clean, p_value_imovel_clean = ttest_ind(imovel_sim_clean, imovel_nao_clean)
print(f'Teste t para posse de imóvel: t_stat={t_stat_imovel_clean}, p_value={p_value_imovel_clean}')

# Teste t de Student para posse de veículo com dados limpos
t_stat_veiculo_clean, p_value_veiculo_clean = ttest_ind(veiculo_sim_clean, veiculo_nao_clean)
print(f'Teste t para posse de veículo: t_stat={t_stat_veiculo_clean}, p_value={p_value_veiculo_clean}')
````

Mesmo após a tentativa de limpeza dos dados, ainda sim não houve diferença significativa.
